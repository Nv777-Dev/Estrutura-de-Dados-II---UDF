# EXERCÍCIO 1 - SOMA DE UM VETOR

```python
# Criar uma lista vazia chamada numeros e essa lista vai guardar os 10 números digitados.
numeros = []


# Cria uma variável chamada soma e ela começa com 0
soma = 0


# O for vai repetir o código 10 vezes e i começa em 0 e vai até 9
for i in range(10):

    #Pede um valor para o usuário
    numero = int(input("Digite um número: "))

    # append adiciona o número digitado ao final da lista
    numeros.append(numero)


# Mostra uma mensagem antes de exibir os números
print("Números digitados:")


# Percorre as 10 posições da lista
for i in range(10):

    # Mostra o número armazenado na posição i
    print(numeros[i])


# Percorre novamente todos os números
for i in range(10):

    # Pega o valor atual de soma e adiciona o número da posição i
    soma = soma + numeros[i]


# Divide a soma por 10 para encontrar a média
media = soma / 10


# Mostra a soma de todos os números
print("Soma:", soma)


# Mostra a média dos números
print("Média:", media)
```

# Exercício 2 — Maior e menor

```python
# Cria uma lista vazia para guardar os números
numeros = []


# Repete 10 vezes para ler os 10 números
for i in range(10):

    # Pede um número inteiro
    numero = int(input("Digite um número: "))

    # Adiciona o número à lista
    numeros.append(numero)


# Faz com que o primeiro número da lista é o maior valor
maior = numeros[0]


# Faz com que o primeiro número é o menor
menor = numeros[0]


# Guarda a posição do maior número
pos_maior = 0


# Guarda a posição do menor número
pos_menor = 0


# Começa na posição 1, porque a posição 0 já foi usada como referência
for i in range(1, 10):

    # Verifica se o número da posição atual é maior que o maior encontrado até agora
    if numeros[i] > maior:

        # Se for maior, ele passa a ser o novo maior
        maior = numeros[i]

        # Guarda a posição desse número
        pos_maior = i


    # Verifica se o número atual é menor que o menor encontrado até agora
    if numeros[i] < menor:

        # Se for menor, ele passa a ser o novo menor
        menor = numeros[i]

        # Guarda a posição do menor
        pos_menor = i


# Mostra o maior valor
print("Maior valor:", maior)


# Mostra a posição do maior valor
print("Posição:", pos_maior)


# Mostra o menor valor
print("Menor valor:", menor)


# Mostra a posição do menor valor
print("Posição:", pos_menor)
```

# EXERCÍCIO 3 - NÚMEROS PARES

```python
# Cria uma lista vazia
numeros = []


# Contador de números pares
quantidade_pares = 0


# Variável que vai armazenar a soma dos números pares
soma_pares = 0


# Repete 20 vezes
for i in range(20):

    # Pede um número inteiro
    numero = int(input("Digite um número: "))

    # Adiciona o número na lista
    numeros.append(numero)


# Mostra uma mensagem antes dos numeros
print("Números pares:")


# Percorre as 20 posições da lista
for i in range(20):

    # O símbolo % calcula o resto da divisão, Se o resto da divisão por 2 for 0, significa que o número é par
    if numeros[i] % 2 == 0:

        # Mostra o número par
        print(numeros[i])

        # Soma 1 ao contador de pares
        quantidade_pares = quantidade_pares + 1

        # Adiciona o número par à soma
        soma_pares = soma_pares + numeros[i]


# Mostra quantos números pares foram encontrados
print("Quantidade de numeros pares:", quantidade_pares)


# Mostra a soma dos números pares
print("Soma dos numeros pares:", soma_pares)
```

# EXERCÍCIO 4 - INVERSÃO DE VETOR

```python
# Cria uma lista vazia
numeros = []


# Repete 10 vezes
for i in range(10):

    # Pede um número inteiro
    numero = int(input("Digite um número: "))

    # Adiciona o número à lista
    numeros.append(numero)


# Mostra uma mensagem
print("Vetor original:")


# Percorre as 10 posições da lista
for i in range(10):

    # Mostra cada número antes da inversão
    print(numeros[i])


# Como temos 10 números, precisamos fazer 5 trocas
for i in range(5):

    # Guarda o número da posição inicial
    temp = numeros[i]

    # Coloca na posição inicial, o número correspondente do final da lista
    numeros[i] = numeros[9 - i]

    # Coloca na posição final, o número que havíamos guardado em temp
    numeros[9 - i] = temp


# Mostra uma mensagem
print("Vetor invertido:")


# Percorre novamente a lista
for i in range(10):

    # Mostra os números depois da inversão
    print(numeros[i])
```

# Exercício 5 — Matriz 3 × 3

```python
# Cria uma lista vazia
matriz = []


# Variável que vai guardar a soma de todos os números da matriz
soma = 0


# Esse for controla as linhas, como a matriz tem 3 linhas, ele repete 3 vezes
for i in range(3):

    # Cria uma lista vazia que representa uma linha da matriz
    linha = []


    # Esse for controla as colunas, Cada linha possui 3 colunas
    for j in range(3):

        # Pede um número inteiro
        numero = int(input("Digite um número: "))

        # Adiciona o número na linha
        linha.append(numero)


    # Depois de preencher a linha, adiciona essa linha à matriz
    matriz.append(linha)


#  Primeiro número da matriz é o maior
maior = matriz[0][0]


# Mostra uma mensagem antes da matriz
print("Matriz:")


# Percorre as linhas da matriz
for i in range(3):

    # Percorre as colunas da matriz
    for j in range(3):

        # Mostra o elemento que está8 na linha i e coluna j
        print(matriz[i][j], end=" ")

        # Adiciona o elemento atual à soma
        soma = soma + matriz[i][j]

        # Verifica se o elemento atual é maior que o maior encontrado
        if matriz[i][j] > maior:

            # Se for, atualiza o maior valor
            maior = matriz[i][j]


    # Pula para a próxima linha depois de mostrar uma linha da matriz
    print()


# Mostra a soma dos elementos
print("Soma dos elementos:", soma)


# Mostra o maior valor encontrado
print("Maior valor:", maior)
```

# EXERCÍCIO 6 - DIAGONAL PRINCIPAL

```python
# Cria uma lista vazia para representar a matriz
matriz = []


# Variável que armazena a soma da diagonal principal
soma = 0


# Percorre as 4 linhas
for i in range(4):

    # Cria uma nova linha
    linha = []


    # Percorre as 4 colunas
    for j in range(4):

        # Pede um número inteiro
        numero = int(input("Digite um número: "))

        # Adiciona o número na linha
        linha.append(numero)


    # Adiciona a linha completa na matriz
    matriz.append(linha)


# Mostra uma mensagem antes da matriz
print("Matriz:")


# Percorre as linhas
for i in range(4):

    # Percorre as colunas
    for j in range(4):

        # Mostra cada elemento
        print(matriz[i][j], end=" ")

    # Pula para a próxima linha
    print()


# Mostra uma mensagem
print("Diagonal principal:")


# Percorre as posições de 0 até 3
for i in range(4):

    # Na diagonal principal,82 linha e coluna possuem o mesmo índice
    print(matriz[i][i])

    # Soma o elemento da diagonal
    soma = soma + matriz[i][i]


# Mostra a soma da diagonal principal
print("Soma da diagonal:", soma)
```

# EXERCÍCIO 7 - MATRIZ DE NOTAS

```python
# Cria uma lista vazia que vai armazenar as notas
notas = []


# O exercício possui 4 alunos
for i in range(4):

    # Cria uma linha para guardar as notas de um aluno
    linha = []


    # Mostra qual aluno está sendo cadastrado
    print("Aluno", i + 1)


    # Cada aluno possui 3 avaliações
    for j in range(3):

        # Pede uma nota, float permite números com casas decimais
        nota = float(input("Digite a nota: "))

        # Adiciona a nota à linha do aluno
        linha.append(nota)


    # Adiciona as três notas desse aluno à matriz
    notas.append(linha)


# Percorre os 4 alunos
for i in range(4):

    # A soma começa novamente em zero para cada aluno
    soma = 0


    # Percorre as 3 notas do aluno atual
    for j in range(3):

        # Soma cada nota
        soma = soma + notas[i][j]


    # Divide a soma por 3 para encontrar a média
    media = soma / 3


    # Mostra qual aluno está sendo apresentado
    print("Aluno", i + 1)


    # Mostra a média
    print("Média: %.2f" % media)
```

# EXERCÍCIO 8 - PRODUTOS

```python
# Cria uma lista vazia que vai armazenar os produtos
produtos = []


# Repete 5 vezes porque precisamos cadastrar 5 produtos
for i in range(5):

    # Mostra qual produto está sendo cadastrado
    print("Produto", i + 1)


    # Lê o nome do produto
    nome = input("Nome: ")


    # Lê o código e transforma em inteiro
    codigo = int(input("Código: "))


    # Lê o preço e transforma em float
    preco = float(input("Preço: "))


    # Lê a quantidade e transforma em inteiro
    quantidade = int(input("Quantidade: "))


    # Cria uma lista com os dados do produto
    produto = [nome, codigo, preco, quantidade]


    # Adiciona esse produto à lista de produtos
    produtos.append(produto)


# Calcula o valor em estoque do primeiro produto
maior_valor = produtos[0][2] * produtos[0][3]


# Consideramos inicialmente que o produto de maior valor está na posição 0
pos_maior = 0

# Mostra os produtos cadastrados com uma mensagem maior para ficar mais facil a vizualização na saida do codigo

print()
print("===================================")
print("       PRODUTOS CADASTRADOS")
print("===================================")

# Percorre os 5 produtos
for i in range(5):

    # Calcula o valor em estoque preço x quantidade
    valor_estoque = produtos[i][2] * produtos[i][3]


    # Mostra o nome
    print("Nome:", produtos[i][0])


    # Mostra o código
    print("Código:", produtos[i][1])


    # Mostra o preço
    print("Preço:", produtos[i][2])


    # Mostra a quantidade
    print("Quantidade:", produtos[i][3])


    # Mostra o valor total em estoque
    print("Valor em estoque:", valor_estoque)


    # Verifica se o valor em estoque atual é maior que o maior encontrado anteriormente
    if valor_estoque > maior_valor:

        # Atualiza o maior valor
        maior_valor = valor_estoque

        # Guarda a posição desse produto
        pos_maior = i

# Mostra os produtos com maior valor em estoque com uma mensagem maior para ficar mais facil a vizualização na saida do codigo
print()
print("===================================")
print("  PRODUTO COM MAIOR VALOR EM ESTOQUE")
print("===================================")

# Mostra o nome do produto que possui o maior valor em estoque
print("Nome:", produtos[pos_maior][0])


# Mostra o valor
print("Valor:", maior_valor)
```

# EXERCÍCIO 9 - ALUNOS

```python
# Cria uma lista para armazenar os alunos
alunos = []


# Contador de alunos aprovados
aprovados = 0


# Contador de alunos reprovados
reprovados = 0


# Repete 5 vezes para cadastrar os 5 alunos
for i in range(5):

    # Mostra qual aluno está sendo cadastrado
    print("Aluno", i + 1)


    # Lê o nome
    nome = input("Nome: ")


    # Lê a idade
    idade = int(input("Idade: "))


    # Lê a primeira nota
    nota1 = float(input("Nota 1: "))


    # Lê a segunda nota
    nota2 = float(input("Nota 2: "))


    # Lê a terceira nota
    nota3 = float(input("Nota 3: "))


    # Soma as três notas e divide por 3 para encontrar a média
    media = (nota1 + nota2 + nota3) / 3


    # Cria uma lista com todos os dados do aluno
    aluno = [
        nome,
        idade,
        nota1,
        nota2,
        nota3,
        media
    ]


    # Adiciona o aluno à lista de alunos
    alunos.append(aluno)


# Consideramos inicialmente que o primeiro aluno possui a maior média
maior_media = alunos[0][5]


# Guarda a posição do aluno com maior média
pos_maior = 0


# Percorre os 5 alunos
for i in range(5):

    # Mostra o nome do aluno
    print("Aluno:", alunos[i][0])


    # Mostra sua média
    print("Média:", alunos[i][5])


    # Verifica se a média é maior ou igual a 7
    if alunos[i][5] >= 7:

        # Se for, o aluno está aprovado
        print("Situação: Aprovado")

        # Aumenta o número de aprovados
        aprovados = aprovados + 1


    # Caso a condição anterior não seja verdadeira
    else:

        # O aluno está reprovado
        print("Situação: Reprovado")

        # Aumenta o número de reprovados
        reprovados = reprovados + 1


    # Verifica se a média do aluno atual é maior que a maior média encontrada
    if alunos[i][5] > maior_media:

        # Atualiza a maior média
        maior_media = alunos[i][5]

        # Guarda a posição do aluno
        pos_maior = i

# Mostra o resumo final dos alunos em uma mensagem maior para melhor visualização

print()
print("===================================")
print("           RESUMO FINAL")
print("===================================")

# Mostra quantos alunos foram aprovados
print("Quantidade de aprovados:", aprovados)


# Mostra quantos alunos foram reprovados
print("Quantidade de reprovados:", reprovados)


# Mostra uma mensagem
print("Aluno com maior média:")


# Mostra o nome do aluno
print("Nome:", alunos[pos_maior][0])


# Mostra sua média
print("Média:", maior_media)
```

# DESAFIO - CADASTRO DE ALUNOS

```python
# Cria uma lista vazia para armazenar os 5 alunos
alunos = []

# Repete 5 vezes porque serão cadastrados 5 alunos
for i in range(5):

    # Mostra qual aluno está sendo cadastrado
    print("Aluno", i + 1)


    # Lê o nome
    nome = input("Nome: ")


    # Lê a idade
    idade = int(input("Idade: "))


    # Cria uma lista vazia que vai guardar as três notas
    notas = []


    # Repete 3 vezes para ler as três notas
    for j in range(3):

        # Lê uma nota
        nota = float(input("Digite a nota: "))

        # Adiciona a nota à lista de notas
        notas.append(nota)


    # Cria a variável que armazenará a soma das notas
    soma = 0


    # Percorre as três notas
    for j in range(3):

        # Adiciona cada nota à soma
        soma = soma + notas[j]


    # Divide a soma por 3 para encontrar a média
    media = soma / 3


    # Cria uma lista com os dados do aluno
    aluno = [
        nome,
        idade,
        notas,
        media
    ]


    # Adiciona esse aluno à lista de alunos
    alunos.append(aluno)

# Mostra uma mensagem
print("Médias dos alunos:")


# Percorre os 5 alunos
for i in range(5):

    # Mostra o nome
    print("Nome:", alunos[i][0])

    # Mostra a média
    print("Média:", alunos[i][3])


# --------------------------------
# MAIOR MÉDIA
# --------------------------------


# Consideramos inicialmente a média do primeiro aluno como a maior
maior_media = alunos[0][3]


# Guarda a posição do aluno que possui a maior média
pos_maior = 0


# Começa no segundo aluno porque o primeiro já foi usado como referência
for i in range(1, 5):

    # Verifica se a média atual é maior que a maior encontrada
    if alunos[i][3] > maior_media:

        # Atualiza a maior média
        maior_media = alunos[i][3]

        # Guarda a posição do aluno
        pos_maior = i


# Mostra uma mensagem
print("Aluno com maior média:")


# Mostra o nome do aluno
print("Nome:", alunos[pos_maior][0])


# Mostra a maior média
print("Média:", maior_media)
```
